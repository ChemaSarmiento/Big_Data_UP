```python
RUTA = "gs://<BUCKET_UNICO>/muestras/profeco.csv"
# El facilitador proporciona una SparkSession llamada spark.
from pyspark.sql import functions as F
df = spark.read.option("header", True).csv(RUTA)
required = {"producto", "precio", "estado"}
assert required <= set(df.columns), "Revisar columnas del CSV preparado"
df = df.withColumn("precio", F.col("precio").cast("double"))
df.select("producto", "precio", "estado").show(5)
```

## 1. Explorar una pregunta
¿En qué estado cambia el precio de un producto? Completa `PRODUCTO` y `ESTADO` con valores que existan en la muestra.


```python
PRODUCTO = "<producto de la muestra>"  # TODO 1
ESTADO = "<estado de la muestra>"      # TODO 2
consulta = df.filter((F.col("producto") == PRODUCTO) & (F.col("estado") == ESTADO))
consulta.select("producto", "precio", "estado").show(10)
```

## 2. Comparar grupos
Completa `GRUPO` con `estado` y `MEDIDA` con `precio`. Antes de ejecutar, predice qué fila tendrá mayor promedio.


```python
GRUPO = "<columna para agrupar>"  # TODO 3
MEDIDA = "<columna numérica>"      # TODO 4
resultado = df.filter(F.col("producto") == PRODUCTO).groupBy(GRUPO).agg(
    F.avg(MEDIDA).alias("precio_promedio"), F.count(MEDIDA).alias("observaciones")
).orderBy(F.desc("precio_promedio"))
resultado.show(10)
```
